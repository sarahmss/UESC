# 2. Configurar registradores de controle e status 

### 2.1 Registrador A (ADCSRA)

• ADEN: ADC Enable 
    ADEN = 0 -> ADC desligado
    ADEN = 1 -> ADC ligado

• ADSC: ADC Start Conversion 
    
    ADSC = 1 -> Inicia cada conversão no modo de conversão única
    
    ADSC = 1 -> Inicia a primeira conversão no modo de conversão contínua

    ADSC = 0 -> Não tem efeito (Quando a conversão  for completa, o bit é setado como zero)

• ADATE: ADC Auto Trigger Enable
    ADATE = 1 -> Acionamento automático ativado (modo contínuo) 
    
        Rising edge do sinal de disparo inicia a conversão AD

• ADIF: ADC Interrupt Flag
    ADC == 1 -> Conversão AD completa e o registrador de dados foi atualizado

• ADIE: ADC Interrupt Enable
    ADIE = 1 -> Ativa a interrupção de final de conversão AD 

• ADPS2:0: ADC Prescaler Select Bits

Se o resultado for ajustado à esquerda e não for necessária uma precisão maior que 8 bits, basta ler ADCH. Caso contrário, ADCL deve ser lido primeiro e, em seguida, ADCH.
